```python
# 37 - find the largest word in string
str37=input("enter a string = ")
words = str37.split()
largest = words[0]
for word in words:
    if len(word) > len(largest):
        largest = word
print(largest)        
```

```python
#38 - find the shortest word in string

str38=input("enter a sentences = ")
words=str38.split()
shortest=words[0]
for word in words:
    if len(word) < len(shortest):
        shortest=word
print(shortest)        
```
