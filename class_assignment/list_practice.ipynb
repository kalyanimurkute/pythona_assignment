```python
#list update operation perform

#normal list
lst=[1,2,3,4, 5]
for i in lst:
    print(i)

#add- append()
print("\n!------using append-----!\n")
lst=input("enter a list seperated by comma = ").split(",")
add=input("enter add element = ")
lst.append(add)
for i in lst:
    print(i)

#insert()
print("\n!-----using insert-----!\n")
print("after using insert")
index_n=int(input("enter a index number = "))
value=int(input("enter a value = "))
lst.insert(index_n,value)
for i in lst:
    print(i)
```

```python
# update

lst=[1,2,3,4, 5]
for i in lst:
    print(i)
#after updating list
print("\n---update the list---\n")
index=int(input("enter a index number = "))
new=int(input("enter a updating value = "))
lst[index]=new
for i in lst:
    print(i)
```

```python
#remove
print("\n!----remove----!\n")
lst=input("enter a element seperated by comma = ").split(",")
print(lst)
remove=input("enter a remove element = ") 
lst.remove(remove) 
for i in lst:
    print(i)   


#pop()
print("\nusing the pop\n")
lst.pop()
for i in lst:
    print(i)

#pop(index)
print("\nusing pop(index)\n")
index=int(input("enter index for remove the element = "))
lst.pop(index)
for i in lst:
    print(i)
```

```python
#mini-project for add student, update student and remove student
'''
1. add student and her all details
2.update student information
3.remove the student specific inforamtion
'''
print("\nstudent Information :\n1.roll_no \n2.name \n3.surname \n4.address")
student=input("enter student inforamtion seperated by comma = ").split(",")
if student=="":
    print("information cannot be empty")
else:
    while True:
        print("\n!-----------------simple project----------------!\n")
        print("1.add the student information")
        print("2.update the student information")
        print("3.remove the student sinformation")
        ch=int(input("enter a your choice = "))
        match ch:
            case 1:
                # add student
                
                print(student)
                element=input("enter add element = ")
                student.append(element)
                for i in student:
                    print(i)
            case 2:
                #update student
                print(student)
                index=int(input("enter a index for update info "))
                add_val=input("enter a update value = ")
                student[index]=add_val
                for i in student:
                    print(i)     
            case 3:
                index=int(input("enter index of remove element = "))
                student.pop(index)
                for i in student:
                    print(i)
            case 4 :
                print("thank you!")
                break
            case _:
                print("invalid choice")
                break
        
```
