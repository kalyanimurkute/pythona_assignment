```python
#mini project = pasword strength checker


'''
1.check minimum password length
2.check presences of upper case letter
3.check presences of loweer case letter
4.check presence of digit
5.check presences of special character
6.display password strength result
'''


passw=input("enter your password = ")
if passw=="":
    print("password cannot be empty")
else:
    while True:
        print("\n------------------Password Strength Checker-------------------\n")
        print("1.check minimum password length")
        print("2.check presences of upper case letter")
        print("3.check presences of loweer case letter")
        print("4.check presence of digit")
        print("5.check presences of special character")
        print("6.display password strength result")
        print("7.exits")
        ch=int(input("enter your choice = "))
        match ch:
            case 1:
                if 6 <=len(passw) <=15:
                    print("Password length is good")
                else:
                    print("please enter a 6 to 14 digit")
            case 2:
                found= False
                for i in passw:
                    if i.isupper():
                        found=True
                        break
                if found:
                    print("upper case letter is present")
                else:
                    print("upper case letter is not present")
            case 3:
                found=False
                for i in passw:
                    if i.islower():
                        found=True
                        break
                if found:
                    print("lower case letter is present")
                else:
                    print("lower case letter is not present")                    
            case 4:
                found=False
                for i in passw:
                    if i.isdigit():
                        found=True
                        break
                if found:
                    print("digit is presents")
                else:
                    print("digit is not presents")
            case 5:
                found=False
                for i in passw:
                    if i in ('*','&','%','$','@','#'):
                        found=True
                        break
                if found:
                    print("special character is present")
                else:
                    print("special character is not present")
            case 6:
                strength=0
                if 6<=len(passw)<=14:
                    strength +=1
                for i in passw:
                    if i.isupper():
                        strength+=1
                        break
                for i in passw:
                    if i.islower():
                        strength +=1
                        break
                for i in passw:
                    if i.isdigit():
                        strength +=1
                        break
                for i in passw:
                    if i in ('*','&','%','$','@','#'):
                        strength +=1
                        break
                if strength == 5:
                    print("your password srength is strong")
                elif strength >=3:
                    print("your password srength is good")
                else:
                    print("your password srength is weak")
            case 7:
                print("thank you..!")
                break
            case _:
                print("invalid choice")
```
