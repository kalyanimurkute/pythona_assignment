```python
#42-calculator operation using match case
num1=int(input("enter the first number = "))
num2=int(input("enter the second number = "))
print("1.addition")
print("2.substraction")
print("1.multiplication")
print("1.division")
ch=int(input("enter your choice = "))
match ch:
    case 1:
        print("addition of two num = ",num1+num2)
    case 2:
        print("substraction of two num = ",num1-num2)
    case 3:
        print("multiplication of two num = ",num1*num2)
    case 4:
        if num2 !=0:
            print("division of two num = ",num1/num2)
        else:
            print("enter the number not zero ")
    case 5:
        print("invalid number .....enter valid number")
```

```python
#50-atm menu  using match case
bal=50000
print("ATM menu")
print("1. check balance")
print("2. deposite")
print("3. withdraw")
print("4. exit")
ch=int(input("enter choice = "))
match ch:
    case 1:
        print("your balance = ",bal)
    case 2:
        amount=int(input("enter a amount = "))
        bal=bal + amount
        print("updated amount =",bal)
    case 3:
        amount=float(input("enter withdrawal amount = "))
        if amount <= bal:
            bal=bal-amount
            print("updated amount = ",bal)
        else:
            print("insufficient balance")
    case 4:
        print("thank you!  visit again")
    case _:
        print("invalid choice")
        
```
