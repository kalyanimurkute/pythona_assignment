```python
#validation method - isalpha

name=input("enter a name = ").title()
if name=="":
    print("name cannot be empty.")
elif not name.isalpha():
    print("enter only alphabet ")
else:
    print(name)
```

```python
#isdigit

mob=input("enter a your mobile number = ")
if mob=="":
    print("mobile number cannot be empty.")
elif  len(mob)!=10:
    print("invalid number.\nplease enter 10 digit of number.")    
elif mob[0] not in ['9','8','7','6']:
    print("not indian number")
else:
    print(mob)
    
```
