```python
# 24 Check Number Is Strong Number or Not.
num=145
sum1=0
temp=num
while temp > 0:
    digit = temp % 10
    fact=1
    for i in range(1,digit + 1):
        fact = fact * i
    sum1=sum1 + fact
    temp = temp // 10
if sum1 == num:
    print("strong")
else:
    print("not")
    
```
