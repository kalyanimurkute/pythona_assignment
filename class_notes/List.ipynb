```python
#adding append()
# how we can take user input inside the list 
# first ways 
lst = []
how_many_element = int(input("Enter the how many element you want to add = "))
for i in range(how_many_element):
    ele = eval(input("Enter the elements = "))
    lst.append(ele)

# display the elements
# for data in lst:
#     print(data)
print(lst)
```

```python
#second way to add element inside the list by using list comprehension 
lst = [int(num) for num in input("Enter the number seprated by comma = ").split(',')]
print(lst)
```

```python
# adding insert(index, value) [10,20,30]
lst = []
how_many_element = int(input("Enter the how many element you want to add = "))
for i in range(how_many_element):
    ele = eval(input("Enter the elements = "))
    lst.append(ele)
    
indx = input("Enter the index number for insert the value = ")
if not indx.isdigit():
    print("Only digit are allowed")
else:
# convert into int
    indx = int(indx)
    val = int(input("Enter the value = "))

    # call the insert()
    lst.insert(indx,val)

    # print the elememt
    print(lst)

```

```python
#update method 
'''
lst[1] = 200
'''
lst = []
how_many_element = int(input("Enter the how many element you want to add = "))
for i in range(how_many_element):
    ele = eval(input("Enter the elements = "))
    lst.append(ele)

indx = int(input("Enter the index number to update the value = "))
if indx > len(lst):
    print("index not present")
else:
    val = eval(input("Enter the value = "))
    # updte code 
    lst[indx] = val 
    print(lst)
```

```python
'''
Delete operations
----------------------------
1. pop() : It will remove the last element from list.
2. pop(index) : It will remove specific element using index
3. remove(value) : It will remove the value if the value present it will remove.
'''

lst1 = [1,2,3,4,5]
val = int(input("Enter the value to remove = "))
if val not in lst1:
    print("Value not found")
else:
# lst1.pop()
# lst1.pop()

# lst1.pop(1)
    lst1.remove(val)
    print(lst1)
```

```python
#index()
lst = []
limit = int(input("Enter how many elements you want = "))
for i in range(limit):
    ele = eval(input("Enter the elements = "))
    lst.append(ele)

val = eval(input("Enter the value to get index number = "))

if val not in lst:
    print(f"The value {val} is not present")
else:
    # call the index method 
    res = lst.index(val)
    print(res)
```

```python
# count()
lst = [23,45,2,3,1,2,3,4,8,9]

val = int(input("Enter the value to print the count = "))
# call the count() 
res = lst.count(val)
if res == 0:
    print(f"The value {val} is not present")
else:
    print(res)
```

```python
# pattern


n=8
for i in range(n):
    for j in range(n):
        if(i==0 or i==n-1 or j==0 or j==n-1):
            print('*',end='')
        elif(i==j or i + j == n-1):
            print('*',end ='')
        else:
            print('',end='')
    print()    
      
```
