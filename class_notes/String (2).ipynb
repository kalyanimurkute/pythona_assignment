```python
# Validations Methods(Boolean): 
# isalpha() 
name = input("Enter the name of student = ")
if name == "":
    print("name cannot be empty")
elif not name.isalpha():
    print("only alphabets allowed")
else:
    print(name)
```

```python
# isdigit()
# indian mobile number startswith digit like 9,8,7,6
mob = input("Enter the mobile number = ")
if mob == "":
    print("mobile cannot be empty")
elif not mob.isdigit():
    print("only digit allowed")
elif len(mob) > 10 or len(mob) < 10:
    print("Invalid number. Plese check again")
elif not mob.startswith("9") and not mob.startswith("8") and not mob.startswith("7"): 
    print("Not indian number")
else:
    print(mob)
```

```python
# isalnum() 
email = input("Enter the email = ")
if email == "": 
    print("email cannot be empty")
elif not email.isalnum():
    print("check email")
else:
    print(email)
```
